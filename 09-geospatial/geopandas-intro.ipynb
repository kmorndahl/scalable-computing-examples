```python
# Read in vector file

comm = gpd.read_file("Alaska_Commercial_Salmon_Boundaries.gpkg")
```

```python
# Explore vector data

comm
comm['geometry'][:5]
comm.crs
comm["geometry"][8]
comm.plot(figsize=(9,9))
```

```python
# Same
comm_3338 = comm.to_crs("EPSG:3338")
comm_3338 = comm.to_crs(ships_meta["crs"])

print(ships_meta["crs"].to_authority()) # Convert from WKT to EPSG code

comm_3338.plot()
```

```python
# Crop data

from shapely.geometry import box

coord_box = box(-159.5, 55, -144.5, 62)
coord_box_df = gpd.GeoDataFrame(
    crs = "EPSG:4326",
    geometry = [coord_box]
).to_crs("EPSG:3338")
```

```python
import rasterio.mask
import numpy as np

# If your raster file is tiled/chunked, the rasterio.mask function will only grab the necessary tiles, it won't read in the whole file
# Does not work that way for COGs which are not tiled
with rasterio.open("Coastal_2020_08.tif") as ship_con:
    shipc_arr, shipc_transform = rasterio.mask.mask(ship_con,
                                                    coord_box_df["geometry"],
                                                    crop=True)
    shipc_meta = ship_con.meta
    # select just the 2-D array (by default a 3-D array is returned even though we only have one band)
    shipc_arr = shipc_arr[0, :, :]
    # turn the no-data values into NaNs.
    shipc_arr[shipc_arr == ship_con.nodata] = np.nan
```

```python
shipc_meta.update({"driver": "Gtiff",
                   "height": shipc_arr.shape[0],
                   "width": shipc_arr.shape[1],
                   "transform": shipc_transform,
                   "compression": "lzw",
                   "nodata": np.nan})
```

```python
comm_clip = gpd.sjoin(comm_3338, 
                      coord_box_df,
                      how = 'inner',
                      predicate = 'within')
```

```python
import rasterio.plot

fig, ax = plt.subplots(figsize = (7, 7))

rasterio.plot.show(shipc_arr,
                   ax = ax,
                   vmin = 0,
                   vmax = 5000,
                   transform = shipc_transform)

comm_clip.plot(ax = ax, facecolor = "none", edgecolor = "red")
```

```python
from rasterio import features

comm_clip["geometry"][40]

# Rasterize a single polygon
# 1 = within polygon boundaries
# 0 = not within polygon boundaries
r40 = features.rasterize(comm_clip["geometry"][40].geoms,
                         out_shape = shipc_arr.shape,
                         transform = shipc_meta["transform"],
                         all_touched = True)
```

```python
fig, ax = plt.subplots(figsize = (7,7))

rasterio.plot.show(r40,
                   ax=ax,
                   vmin = 0,
                   vmax = 1,
                   transform = shipc_meta['transform'])

comm_clip.plot(ax = ax,
               facecolor = "none",
               edgecolor = "red")
```

```python
np.unique(r40)
```

```python
# Gets x/y index locations of where r40 is equal to one
r40_index = np.where(r40 == 1)
r40_index
```

```python
# Get total meters travelled within fishing district r40

np.nansum(shipc_arr[r40_index]) # Grabs indexes within district r40 and sums all 'shipc_arr' values
```

```python
# Get total meters travelled within all fishing districts

# Add ID column
comm_clip["id"] = range(0, len(comm_clip))

# Loop through features in vector file
distance_dict = {}
for geom, idx in zip(comm_clip["geometry"], comm_clip["id"]):
    # Rasterize current feature
    rasterized = features.rasterize(geom.geoms,
                                    out_shape = shipc_arr.shape,
                                    transform = shipc_meta["transform"],
                                    all_touched = True)
    # Get indexes associated with current rasterized feature
    r_index = np.where(rasterized == 1)
    # Calculate total distance within feature
    total_distance = np.nansum(shipc_arr[r_index])
    # Assign to dictionary
    distance_dict[idx] = total_distance

```

```python
distance_dict

import pandas as pd

distance_df = pd.DataFrame.from_dict(distance_dict,
                                      orient = "index",
                                      columns = ["distance"])

distance_df["id"] = distance_df.index
distance_df["distance"] = distance_df["distance"] / 1000

distance_df
```

```python
res_full = comm_clip.merge(distance_df,
                           on = "id",
                           how = "inner")

res_full
```

```python
import matplotlib.ticker
fig, ax = plt.subplots(figsize=(7, 7))

ax = res_full.plot(column = "distance", legend = True, ax = ax)
fig = ax.figure
label_format = '{:,.0f}'
cb_ax = fig.axes[1]
ticks_loc = cb_ax.get_yticks().tolist()
cb_ax.yaxis.set_major_locator(matplotlib.ticker.FixedLocator(ticks_loc))
cb_ax.set_yticklabels([label_format.format(x) for x in ticks_loc])
ax.set_axis_off()
ax.set_title("Distance Traveled by Ships in Kilometers")
plt.show()
```

```python
reg_area_sum = res_full.dissolve(by = "REGISTRATION_AREA_NAME",
                                 aggfunc = "sum",
                                 numeric_only = True)
reg_area_sum
```
